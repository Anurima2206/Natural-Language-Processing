```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('train.txt',sep=';',header=None,names=['text','emotion'])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
unique_emotions=df['emotion'].unique()
emotion_numbers={}
i=0
for(emo) in unique_emotions:
    emotion_numbers[emo]=i
    i+=1
df['emotion']=df['emotion'].map(emotion_numbers)
```

```python
df.head()
```

```python
# make lowercase
df['text']=df['text'].apply(lambda x:x.lower())
```

```python
#remove punctuations
import string
def remove_punc(text):
  return text.translate(str.maketrans('','',string.punctuation))
df['text']=df['text'].apply(remove_punc)
```

```python
from IPython.utils import text
#remove numbers
def remove_numbers(text):
  new=""
  for i in text:
    if not i.isdigit():
      new=new+i
  return new
df['text']=df['text'].apply(remove_numbers)
```

```python
#remove emojis
def remove_emojis(text):
  new=""
  for i in text:
    if i.isascii():
      new=new+i
  return new
df['text']=df['text'].apply(remove_emojis)
```

```python
#remove stopwords
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
stop_words=set(stopwords.words('english'))
def remove_stopwords(text):
  word_tokens=word_tokenize(text)
  cleaned=[i for i in word_tokens if i not in stop_words]
  return ' '.join(cleaned)

df['text']=df['text'].apply(remove_stopwords)
```

One hot encoding is not used because:
i. It forms sparse matrix
ii. No semantic meaning covered
iii. Out of Vocabulary does not support test words not included in vocabulary
iv. Has different dimensions for different documents


```python
#vectorization
#EXAMPLE OF BAG OF WORDS VECTORIZATION
from sklearn.feature_extraction.text import CountVectorizer

documents=[
    "I love pizza",
    "Pizza is the best",
    "I love pasta",
    "Pasta is great"
]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(documents)
print("Vocabulary:",vectorizer.get_feature_names_out())
print("\nBoW Matrix:\n",X.toarray())
```

Bag of words is used but has some

Disadvantages because:
i. It forms sparse matrix
ii. No semantic meaning covered
iii. Out of Vocabulary does not support test words not included in vocabulary
iv. If two documents are for eg. Cricket is very good. Cricket is not good...here out of 4dimensions...only 1 is diff, rest 3 same...so shows similarity despite opposite meanings. SO bag of ngrams is used.

Advantages:
i. Comparitively semantic meaning more covered than one hot encoding as uses frequency.
ii. Easy to implement.
iii. Provides encodings of same dimensions throughout.

```python
#EXAMPLE OF BAG OF NGRAMS VECTORIZATION
from sklearn.feature_extraction.text import CountVectorizer

documents=[
    "I love pizza",
    "Pizza is the best",
    "I love pasta",
    "Pasta is great"
]

vectorizer = CountVectorizer(ngram_range=(1,2))#means group of 1, group of 2 formed
X = vectorizer.fit_transform(documents)
print("Vocabulary:",vectorizer.get_feature_names_out())
print("\nBoW Matrix:\n",X.toarray())
```

Bag of ngrams is used

Advantages:
i.More semantic meaning covered as for example two documents are for eg. Cricket is very good. Cricket is not good...here out of 5dimensions...only 1 is same, rest 3 diff...so shows different meanings.
ii. Rest adv of bag of words.

Disadvantages:
i. Dimensions become too large.
ii. Out of Vocabulary does not support test words not included in vocabulary

```python
#EXAMPLE OF BAG OF TFIDF VECTORIZATION
from sklearn.feature_extraction.text import TfidfVectorizer

documents=[
    "I love pizza",
    "Pizza is the best",
    "I love pasta",
    "Pasta is great"
]

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(documents)
print("Vocabulary:",vectorizer.get_feature_names_out())
print("\nBoW Matrix:\n",X.toarray())
```

TF-Term Frequency => No.of occurence of term in document / Total no. of terms in document

IDF-Inverse Document Frequency => Total no. of docum in corpus/ No. of documents with term in it

Vector of a term= TF x IDF

Less occuring terms have more IDF and lower TF.

Advantages:
i.Information Retrieval since more semantic meanings captured.

Disadvantages:
i. Sparsity
ii. Out of Vocabulary does not support test words not included in vocabulary

```python
#returning back to emotion dataset for nlp
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df['text'],df['emotion'],test_size=0.2,random_state=42)
```

```python
df.shape
```

```python
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
#using countvectorizer
bow_vectorizer=CountVectorizer()
X_train_bow=bow_vectorizer.fit_transform(X_train)
X_test_bow=bow_vectorizer.transform(X_test)
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
model=MultinomialNB()
model.fit(X_train_bow,y_train)
pred_bow=model.predict(X_test_bow)
accuracy_score(y_test,pred_bow)
```

```python
#using tfid vectorizer
tfif_vectorizer=TfidfVectorizer()
X_train_tfif=tfif_vectorizer.fit_transform(X_train)
X_test_tfif=tfif_vectorizer.transform(X_test)
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
model=MultinomialNB()
model.fit(X_train_tfif,y_train)
pred_tfif=model.predict(X_test_tfif)
accuracy_score(y_test,pred_tfif)
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
model.fit(X_train_bow,y_train)
pred_bow=model.predict(X_test_bow)
accuracy_score(y_test,pred_bow)
```
